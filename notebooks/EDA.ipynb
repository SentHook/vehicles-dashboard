# Verificación del entorno de desarrollo.
Importación de librerías necesarías para el desarrollo y verificación del entorno.

```python
import sys
from pathlib import Path
import pandas as pd
import plotly.graph_objects as go

print("Interprete:", sys.executable)
print("Directorio actual:", Path.cwd())
print("Dataset disponible:", Path("../vehicles_us.csv").is_file())
```

# Carga del dataset.
Se cargan los datos y visualización de datos.

```python
car_data = pd.read_csv("../vehicles_us.csv")

car_data.head()
```

```python
print(car_data.head())
print()
car_data.info()
print()
print(car_data.isna().sum())
print()
print(car_data.describe())
```

# Gráfico de Histograma.

```python
fig = go.Figure(
    data=[go.Histogram(x=car_data["odometer"])]
)

fig.update_layout(
    title="Distribución del kilometraje",
    xaxis_title="Kilometraje (milla)",
    yaxis_title="Cantidad de vehículos",
)

fig.show()
```

# Gráfico de dispersión

```python
fig = go.Figure(
    data=[go.Scatter(
        x=car_data["odometer"],
        y=car_data["price"],
        mode="markers",
    )
         ]
)

fig.update_layout(
    title="Relación entre kilometraje y precio",
    xaxis_title="Kilometraje (millas)",
    yaxis_title="Precio (USD)",
)

fig.show()
```
